

# P(class | features) = P(features | class) × P(class) / P(features)

 “naive” part is the assumption that all features are independent of each other, given the class. So instead of estimating a huge joint probability, it just multiplies simple per-feature probabilities:

P(x₁, x₂, …, xₙ | C) = P(x₁|C) × P(x₂|C) × … × P(xₙ|C)

## Variants of Naive Bayes

| Variant | Best for | Data type |
|---|---|---|
| `GaussianNB` | Continuous features | Assumes a normal distribution per class |
| `MultinomialNB` | Word counts, TF-IDF | Text |
| `BernoulliNB` | Binary features | Word present or absent |
| `ComplementNB` | Imbalanced text data | Text |
| `CategoricalNB` | Categorical features | Categories |

```python
df=sns.load_dataset('penguins')
```

```python
df.head(3)
```

```python
df.isnull().sum()
```

```python
df=df.dropna()
```

```python
df=df.drop_duplicates()
```

```python
df.info()
```

```python
# Convert category into the numeric codes
mappings={}
for col in ['species','island','sex']:
    df[col]=df[col].astype("category")
    mappings[col] = dict(enumerate(df[col].cat.categories))  # 1) save mapping first
    df[col] = df[col].cat.codes 
```

```python
X=df.drop(columns=['species'])
y=df['species']
```

```python
X.head(4)
```

```python
X.shape,y.shape
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=10)
```

```python
from sklearn.naive_bayes import GaussianNB
model=GaussianNB()
```

```python
model.fit(X_train,y_train)
```

```python
model=GaussianNB().fit(X_train,y_train)
print("accuracy:",model.score(X_test,y_test))
```
