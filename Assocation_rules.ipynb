```python
# Split the genres columns for a list
```

```python
df[['Genre_1','Genre_2','Genre_3']]=df['Genres'].str.split(',',n=2,expand=True)
```

```python
# Remove Genres columns
df.drop(columns="Genres",inplace=True)
```

```python
df.head(2)
```

```python
# converting into list
```

```python
df['Genres_list']=df[['Genre_1','Genre_2','Genre_3']].values.tolist()
```

```python
df.head(2)
```

```python
df['Genres_list']=df['Genres_list'].apply(lambda x:[i.strip() for i in x if isinstance(i,str)])
```

```python
df.head(2)
```

```python
from mlxtend.preprocessing import TransactionEncoder
```

```python
te=TransactionEncoder()
te_ary=te.fit(df['Genres_list']).transform(df['Genres_list'])
genre_df=pd.DataFrame(te_ary,columns=te.columns_)
```

```python
# Apriori
```

```python
from mlxtend.frequent_patterns import apriori
```

```python
frequent_genres=apriori(genre_df,min_support=0.01,use_colnames=True)
frequent_genres.sort_values(by='support',ascending=False).head()
```

```python
from mlxtend.frequent_patterns import association_rules
rules=association_rules(frequent_genres,metric='lift',min_threshold=1)
rules[['antecedents','consequents','support','confidence','lift']].head()
```
