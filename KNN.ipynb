# KNN (K-Nearest Neighbors)
The intuition

“Tell me who your neighbors are, and I’ll tell you who you are.”

KNN predicts by looking at the most similar past examples. It’s simple, flexible, and explainable, but it needs scaled data, a well-tuned K, and a dataset that’s not too big or too high-dimensional.

```python
import numpy as np
import seaborn as sns
```

```python
df=sns.load_dataset('titanic')
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.impute import KNNImputer

# 1. Columns used to find the "similar" rows (numeric only, include age, exclude target)
num_cols = ["age", "fare", "sibsp", "parch"]      # <- change to your columns

# 2. Keep a flag of which ages were missing (useful later)
df["age_was_missing"] = df["age"].isna().astype(int)

# 3. Scale (NaNs are ignored while scaling and stay as NaN)
scaler = StandardScaler()
scaled = scaler.fit_transform(df[num_cols])

# 4. KNN imputation
imputer = KNNImputer(n_neighbors=5, weights="distance")
imputed_scaled = imputer.fit_transform(scaled)

# 5. Convert back to original units
imputed = scaler.inverse_transform(imputed_scaled)

# 6. Put only the age column back
df["age"] = imputed[:, num_cols.index("age")].round()

print(df["age"].isna().sum())    # should be 0
```
