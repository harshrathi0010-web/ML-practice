# Word Embeddings 
- In natural language processing, word embedding is a term used for the representation of words for text analysis, typically in the form of a real-valued vector that encodes the meaning of the word such that the words that are closer in the vector space are expected to be similar in meaning.

# Word2vec
- It is a technique for learning word embeddings from a large body of text using a shallow (two-layer) neural network. It was introduced by Tomas Mikolov and colleagues at Google in 2013.
- Core idea: a word’s meaning is determined by the words that appear around it. The model trains on text so that words used in similar contexts get similar vectors.

- Two architectures:

- CBOW (Continuous Bag of Words): predicts a target word from its surrounding context words. It’s faster and works well for frequent words.
- Skip-gram: does the reverse, predicting the surrounding context words from a target word. It’s slower but handles rare words better.

# The Core Intuition

- Word2vec is built on the distributional hypothesis: “You shall know a word by the company it keeps.” Words that appear in similar contexts tend to have similar meanings. “Cat” and “dog” both appear near “pet,” “feed,” and “vet,” so they should get similar vector

# CBOW in Detail
- Take the one-hot vectors of all context words.
- Look up each one’s embedding from W.
- Average them to get one hidden vector h.
- Compute scores for every word: u = W'ᵀ · h
- Apply softmax to get probabilities over the vocabulary.
- Compare with the true center word using cross-entropy loss, then backpropagate.

# Skip-gram in Detail
- Take the one-hot of the center word and look up its embedding h.
- Predict the probability of each context word: P(context | center) = exp(v'ᶜᵒⁿᵗᵉˣᵗ · vᶜᵉⁿᵗᵉʳ) / Σ over all words exp(v’ʷ · vᶜᵉⁿᵗᵉʳ)
- Objective: maximize the average log probability of all context words across the corpus.
