#  **Regular Expression (regex)**--> It is a pattern language for finding, matching, extracting or replacing text. In NLP it is the main tool for rule-based text cleaning. In Python it lives in the built-in re module, so there is nothing to install.

# Removing Stop words
- a ,the ,of and ,are my etc are called stopwards
- they contribute to the sentence but did not have value

```python
from nltk.corpus import stopwords
stopwords.words('spanish')
```

```python
def remove_textwords(text):
    new_text=[]

    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)

        x=new_text[:]
        new_text.clear()
        return ' '.join(x)
        
    
```

```python
df['review']=df['review'].apply(remove_textwords)
```

# Handling emoji's

```python
import re

def remove_emoji(text):
    emoji_pattern = re.compile("["
                               u"\U0001F600-\U0001F64F"  # emoticons
                               u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                               u"\U0001F680-\U0001F6FF"  # transport & map symbols
                               u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                               u"\U00002702-\U000027B0"
                               u"\U000024C2-\U0001F251"
                               "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
remove_emoji("Loved the movie. It was 😘😘")
```

```python
remove_emoji("Lmao 😂😂")
```

# Stemming
- Stemming chops the ends off words using fixed rules to get a root form, called the stem. It doesn’t check a dictionary, so the stem may not be a real word.
- studies  → studi
- running  → run
- flies    → fli

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
    
```

```python
sample="walk walking walked walks"
stem_words(sample)
```

# Lemmatization
- Lemmatization converts a word to its dictionary base form, called the lemma. It uses a vocabulary and the word’s part of speech (noun, verb, adjective), so the result is always a real word.
- studies  → study
- running  → run   (as a verb)
- better   → good  (as an adjective)
- wolves   → wolf

```python
import nltk
nltk.download('wordnet')
from nltk.stem import PorterStemmer,WordNetLemmatizer
```

```python
ps=PorterStemmer()
wl=WordNetLemmatizer()
```

```python
# Stemming
print(ps.stem("running"))
print(ps.stem("studies"))
print(ps.stem("flies"))
```

```python
# Lemmatization
print(wl.lemmatize("studies"))               # study
print(wl.lemmatize("wolves"))                # wolf
print(wl.lemmatize("running"))               # running  (default is noun, so it stays)
print(wl.lemmatize("running", pos="v"))      # run
print(wl.lemmatize("better", pos="a"))       # good
```
