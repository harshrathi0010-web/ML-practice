# NLP

 **NLP (Natural Language Processing)** is the field of making computers read, understand, and generate human language
 - NLP converts text into numbers that keep meaning, then uses those numbers to predict or generate something useful.

# **Why do we use it?**
- Scale: a human reads maybe 100 emails an hour, while a model processes millions.
- Unlocking unstructured data: most business data (emails, reviews, tickets, contracts) is text, and without NLP it sits unused.
- Consistency and speed: the same rules every time, 24/7.
- Better interfaces: users can type or talk naturally instead of learning a query language.

#  Where to use it
- Support automation: ticket routing, intent detection, chatbots
- Search and recommendations: semantic search, RAG over company documents
- Information extraction: resumes, invoices, contracts, medical notes
- Voice of customer: sentiment and topic analysis on reviews and surveys
- Safety: spam, fraud, toxicity, and content moderation
- Productivity: summarization, translation, drafting, code assistance

# Best way to use it

-Start with the simplest thing that could work, and move up only when you can show it’s not enough:

- Step	Approach	Move up when...
- 1	Regex or rules	Patterns are too varied
- 2	TF-IDF + Logistic Regression (a baseline)	Accuracy plateaus
- 3	Fine-tuned small encoder (BERT-family)	You need deeper understanding
- 4	LLM with prompting + RAG	You need generation or little labeled data
- 5	Fine-tuned LLM (LoRA)	Prompting can’t hit quality, cost, or latency targets

# **challenge in NLP**
The major challenges in NLP include ambiguity, sarcasm/irony detection, coreference resolution, named entity recognition in context, low‑resource languages, noisy text, bias in models, multilingual processing, long‑range context handling, and hallucination in large language models .
