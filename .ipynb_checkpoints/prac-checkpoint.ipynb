```python
import numpy as np
```

```python
from sklearn.linear_model import LinearRegression
```

```python
age=np.array([[24],[56],[90]])
weight=np.array([[60],[89],[45]])
model=LinearRegression().fit(age,weight)
```

```python
from sklearn.cluster import KMeans

# age and salary (in thousands), no output column
people = np.array([[23, 90], [24, 95], [45, 120], [50, 130], [35, 30], [38, 35]])
kmeans = KMeans(n_clusters=3, n_init=10, random_state=0)
groups = kmeans.fit_predict(people)          # fit(X) only, then a group per row
```

```python
print("predicted weight for a new age of 23:", (model.predict([[23]])[0], 1))
print("cluster of each person:", groups)
print("number of clusters found:", len(set(groups)))
```

```python
print(kmeans.cluster_centers_)
```
