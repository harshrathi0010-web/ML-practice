# Time Series Analysis (TSA)

- TSA isn’t one algorithm. It’s a family of methods for data where time order matters.
-  The core idea is that today’s value depends on what happened before.

# Where should you use it?
- Sales and demand forecasting (retail, e-commerce)
- Inventory and supply planning
- Website or app traffic
- Electricity, energy and weather data
- Sensor and IoT monitoring (machine failure warnings)
- Finance, such as revenue and cash flow

# where to avoid
- very little history
- mostly random data
- big sudden change
- future is driven by the outside factors
- snapshot data with no axis

# Best way to use it
- Plot first -> Your eyes catch trend, seasonality and outliers before any model does.
- Clean the time axis -> Set a datetime index, fix the frequency, and handle missing dates.
- Decompose ->  the series into trend + seasonality + residual.
- Check stationarity -> (ADF test) and use differencing if needed. Look at the ACF/PACF plots.
- Build a baseline first -> Use naive (“tomorrow = today”) or seasonal naive (“next Monday = last Monday”). If your fancy model can’t beat these, it’s useless.
- Split chronologically -> with no shuffling, and validate with walk-forward.
- Climb the model ladder slowly-> Moving average / Exponential Smoothing → ARIMA/SARIMA → Prophet → ML with lag features (XGBoost) → LSTM/Transformers only if truly needed.
- Evaluate-> with MAE, RMSE and sMAPE, always compared to the baseline.
- Avoid data leakage -> Never let future information leak into training, for example by fitting a scaler on the full dataset.
- Retrain regularly- > in production and monitor for drift, because patterns change over time.
