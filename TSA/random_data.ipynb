```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

np.random.seed(42)

# 5 years of monthly data
dates = pd.date_range("2019-01-01", periods=60, freq="MS")

trend = np.linspace(100, 160, 60)                   # slowly going up
seasonal = 15 * np.sin(2 * np.pi * np.arange(60) / 12)  # repeats every 12 months
noise = np.random.normal(0, 4, 60)                  # random wobble

sales = pd.Series(trend + seasonal + noise, index=dates, name="sales")

# Step 1: always plot first
sales.plot(figsize=(10, 4), title="Monthly Sales")
plt.show()

# Step 2: split into components
result = seasonal_decompose(sales, model="additive", period=12)
result.plot()
plt.show()
```
