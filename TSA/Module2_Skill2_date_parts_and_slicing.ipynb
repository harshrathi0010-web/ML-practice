# Module 2, Skill 2: Date parts and date slicing
Run these cells in the notebook where `stock_data` (your AAPL download) already exists.

```python
import pandas as pd

# Close as a simple Series (yfinance gives AAPL as a 2nd column level)
close = stock_data['Close'].squeeze()

print(close.index.dtype)                    # datetime64[...]
print(close.index.is_monotonic_increasing)  # True
```

## 1. Date parts

```python
print(close.index.year[:3])        # year
print(close.index.month[:3])       # 1 to 12
print(close.index.day_name()[:3])  # Monday, Tuesday, ...
print(close.index.dayofweek[:3])   # 0 = Monday ... 4 = Friday
print(close.index.quarter[:3])     # 1 to 4

parts = pd.DataFrame({
    'close':     close,
    'year':      close.index.year,
    'month':     close.index.month,
    'weekday':   close.index.day_name(),
    'dayofweek': close.index.dayofweek,
    'quarter':   close.index.quarter,
})
parts.head()
```

## 2. Slicing by date (use .loc; both ends are included)

```python
march = close.loc['2024-03']                      # everything in March 2024
print("March trading days:", len(march))

march_range = close.loc['2024-03-01':'2024-03-31']  # same thing, as a date range
print("Range slice rows  :", len(march_range))

q4 = close.loc['2024-10':'2024-12']               # Oct to Dec
print("Q4 rows:", len(q4))

# The start date does not have to exist in the data (Jan 1 is a holiday)
print("Jan 1-5 rows:", len(close.loc['2024-01-01':'2024-01-05']))
```

## 3. Filtering with conditions (note the & and the brackets)

```python
all_march   = close[close.index.month == 3]       # all Marches
mondays     = close[close.index.dayofweek == 0]   # all Mondays
summer      = close[(close.index.month >= 6) & (close.index.month <= 8)]  # Jun to Aug

print("March rows :", len(all_march))
print("Mondays    :", len(mondays))
print("Jun-Aug    :", len(summer))
```

## 4. Looking for patterns with groupby
Use **returns**, not raw prices, so the upward trend does not get mixed in.

```python
returns = close.pct_change()

order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
avg_return_by_weekday = returns.groupby(returns.index.day_name()).mean().reindex(order)
print(avg_return_by_weekday)
```
