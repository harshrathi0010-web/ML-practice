# Comprehensive Guide to Time Series Forecasting Models

This reference guide summarizes the foundational univariate and multivariate time series forecasting models, their equations, requirements, and use cases.

---

## 1. Classification Overview

| Category | Description | Models |
| :--- | :--- | :--- |
| **Univariate Models** | Models that predict a single variable using solely its own historical values and shocks. | AR, MA, ARMA, ARIMA, SARIMA |
| **Multivariate Models** | Models that predict two or more interrelated series simultaneously, capturing cross-series feedback. | VAR, VMA, VARMA, VARIMA |

---

## 2. Univariate Forecasting Models

### 1. AR — Autoregressive Model: $AR(p)$
* **Concept:** Assumes the current value depends linearly on its own past values (lags).
* **Mathematical Equation:**
  $$Y_t = c + \phi_1 Y_{t-1} + \phi_2 Y_{t-2} + \dots + \phi_p Y_{t-p} + \epsilon_t$$
* **Parameters:** $p$ = order of autoregression (number of lagged observations).
* **Stationarity Requirement:** **Must be stationary**.
* **Intuition:** Today's temperature is strongly correlated with yesterday's temperature.

---

### 2. MA — Moving Average Model: $MA(q)$
*(Note: Distinct from rolling average smoothing; this models random shocks).*
* **Concept:** Assumes the current value depends linearly on past forecast errors / white noise shocks.
* **Mathematical Equation:**
  $$Y_t = \mu + \epsilon_t + \theta_1 \epsilon_{t-1} + \theta_2 \epsilon_{t-2} + \dots + \theta_q \epsilon_{t-q}$$
* **Parameters:** $q$ = order of moving average (number of lagged error terms).
* **Stationarity Requirement:** **Must be stationary**.
* **Intuition:** A sudden news event (shock) creates ripple effects that decay after a few periods.

---

### 3. ARMA — Autoregressive Moving Average: $ARMA(p, q)$
* **Concept:** Combines both past values ($AR$) and past error shocks ($MA$) in a single model.
* **Mathematical Equation:**
  $$Y_t = c + \sum_{i=1}^p \phi_i Y_{t-i} + \epsilon_t + \sum_{j=1}^q \theta_j \epsilon_{t-j}$$
* **Parameters:** $p$ (lagged observations), $q$ (lagged errors).
* **Stationarity Requirement:** **Must be stationary**.
* **Why use it?** Often achieves a more parsimonious model (fewer parameters) than high-order pure AR or pure MA models.

---

### 4. ARIMA — Autoregressive Integrated Moving Average: $ARIMA(p, d, q)$
* **Concept:** Extends ARMA to non-stationary series by applying differencing ($d$ times) to remove trends before fitting.
* **Mathematical Representation:**
  $$\phi(B)(1 - B)^d Y_t = c + \theta(B)\epsilon_t$$
* **Parameters:**
  * $p$: AR order
  * $d$: Degree of differencing needed to achieve stationarity ($\Delta Y_t = Y_t - Y_{t-1}$)
  * $q$: MA order
* **Stationarity Requirement:** Works on **non-stationary series with trends** (e.g., stock price levels).

---

### 5. SARIMA — Seasonal ARIMA: $SARIMA(p, d, q)(P, D, Q)_s$
* **Concept:** Extends ARIMA by modeling both non-seasonal and repeating seasonal patterns.
* **Parameters:**
  * Non-seasonal orders: $(p, d, q)$
  * Seasonal orders: $(P, D, Q)$
  * $s$: Seasonality span/frequency (e.g., $s=12$ for monthly data, $s=4$ for quarterly, $s=7$ for daily/weekly).
* **Stationarity Requirement:** Works on **non-stationary series exhibiting trend and seasonality** (e.g., retail sales during holiday seasons).

---

## 3. Multivariate Forecasting Models (Vector Models)

### 6. VAR — Vector Autoregressive: $VAR(p)$
* **Concept:** Multivariate generalization of $AR(p)$. Each variable is modeled as a function of its own past values and the past values of every other variable in the system.
* **Example (Two-variable system $X_t, Y_t$ with lag 1):**
  $$X_t = c_1 + a_{11} X_{t-1} + a_{12} Y_{t-1} + \epsilon_{1,t}$$
  $$Y_t = c_2 + a_{21} X_{t-1} + a_{22} Y_{t-1} + \epsilon_{2,t}$$
* **Stationarity Requirement:** All individual series **must be stationary**.

---

### 7. VMA — Vector Moving Average: $VMA(q)$
* **Concept:** Multivariate generalization of $MA(q)$. Models a vector of variables as linear functions of past shock vectors (residuals).
* **Stationarity Requirement:** All series **must be stationary**.

---

### 8. VARMA — Vector Autoregressive Moving Average: $VARMA(p, q)$
* **Concept:** Combines VAR and VMA. Captures both past series dependencies and joint error dynamics across all variables simultaneously.
* **Stationarity Requirement:** All series **must be stationary**.

---

### 9. VARIMA — Vector Autoregressive Integrated Moving Average: $VARIMA(p, d, q)$
* **Concept:** Extends VARMA to handle non-stationary series. Differences the variables $d$ times to make them stationary, then estimates a VARMA model.
* **Stationarity Requirement:** Handles **non-stationary series with trends**.
*(Note: If series share a long-run equilibrium relationship, a Vector Error Correction Model (VECM) is preferred over VARIMA).*


# 2. Univariate Models
1. AR(p): Autoregressive - depends on its own past values (requires stationarity).
2. MA(q): Moving Average - depends on past forecast errors/shocks (requires stationarity).
3. ARMA(p, q): Combines AR and MA components (requires stationarity).
4. ARIMA(p, d, q): Integrates differencing (d) to handle non-stationary trending series.
5. SARIMA(p, d, q)(P, D, Q)s: Extends ARIMA to handle seasonal periodic patterns.

## 3. Multivariate Models
6. VAR(p): Vector AR - each variable depends on lags of itself and all other variables.
7. VMA(q): Vector MA - system modeled on past multidimensional shock vectors.
8. VARMA(p, q): Combines VAR and VMA for multivariate systems.
9. VARIMA(p, d, q): Differences multivariate trending series before fitting VARMA.

# 📈 Granger Causality Test

The **Granger Causality Test** is a statistical hypothesis test used to determine whether one time series is useful in forecasting another. 

> **Core Intuition:** *"Does knowing past values of $X$ help predict future values of $Y$ better than just knowing past values of $Y$ alone?"*

---

## 1. Important Caveat: Predictive Precedence $\neq$ True Causality
* Granger causality measures **temporal precedence and forecasting capability**, not true physical cause-and-effect.
* **Classic Counterexample:** Roosters crow right before sunrise every day. Past rooster crowing helps predict sunrise mathematically, so $X$ Granger-causes $Y$, but roosters do not cause the sun to rise.

---

## 2. Mathematical Framework

The test compares two autoregressive models:

### Model 1: Restricted Model (Predicting $Y$ using only past $Y$)
$$Y_t = c_1 + \sum_{i=1}^p \alpha_i Y_{t-i} + \epsilon_t$$
