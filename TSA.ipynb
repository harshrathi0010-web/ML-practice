```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
dates=pd.date_range(
    start="2025-01-01",
    periods=365,
    freq="D"

)
dates
```

```python
time=np.arange(365)
time[:30]
```

```python
trend=time*0.30
trend[:30]
```

```python
seasonality=20*np.sin(2*np.pi*time/7)
seasonality[:30]
```

```python
noise=np.random.normal(0,35,365)
noise[:30]
```

```python
sales=100 + trend + seasonality+ noise
sales[:30]
```

```python
plt.figure(figsize=(8,3))
plt.plot(dates,sales)
plt.xlabel("Date")
plt.ylabel("Amount in USD$")
plt.title("Amount in USD$ v/s Datee")
plt.show()
```

```python
df=pd.DataFrame({
    'Dates':dates,
    'Price':sales
})
```
