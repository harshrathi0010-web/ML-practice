# Bagging 
- used when the model is overfit (high variance) and
# Boosting
- used when the model is underfit low bias 

# Bagging (Bootstrap Aggregating)

- Create many random samples of the data with replacement (bootstrap samples).
- Train one model on each sample, in parallel and independently.
- Combine them by averaging (regression) or voting (classification).
- Best with high-variance models like deep trees. Averaging cancels out their individual noise.

# Random Forest = Bagging + extra randomness

- It is bagging with decision trees, but each split only considers a random subset of features.
- This makes the trees less correlated with each other, so averaging works much better.
- Each tree doesn’t see about 37% of the rows, so you get a free validation score (oob_score).

# Boosting

- Models are trained sequentially, and each new model focuses on the mistakes of the previous ones.
- Usually shallow trees (weak learners) are used, and the final prediction is a weighted sum of all of them.
- Main versions: AdaBoost (re-weights misclassified samples), Gradient Boosting (fits the residual errors), and the fast implementations XGBoost, - ---- LightGBM and CatBoost.
- Mostly reduces bias, so it can reach very high accuracy.
